```python


import unittest
import sys
import random
from datetime import datetime, timedelta

class IntelligentNewsAgent:
    """
    An intelligent agent that processes, summarizes, and categorizes news articles.
    Uses mock data for reliable demonstration without API dependencies.
    """

    def __init__(self):
        """
        Initialize the news agent with mock data configuration.
        """
        self.language = "en"
        self.summary_sentences = 3

        # Define categories and their keywords for classification
        self.categories = {
            'Technology': [
                'python', 'ai', 'artificial intelligence', 'software', 'computer',
                'tech', 'algorithm', 'iphone', 'android', 'microsoft', 'google',
                'meta', 'code', 'developer', 'app', 'programming', 'data', 'cloud',
                'cybersecurity', 'blockchain', 'vr', 'ar', 'machine learning',
                'neural network', 'deep learning', 'chatgpt', 'llm', 'transformer'
            ],
            'Business': [
                'business', 'market', 'economy', 'stock', 'investment', 'profit',
                'company', 'ceo', 'merger', 'startup', 'venture', 'capital',
                'finance', 'economic', 'trade', 'revenue', 'earnings', 'market',
                'bank', 'financial', 'acquisition', 'industry', 'corporate',
                'quarterly', 'revenue', 'growth', 'strategy', 'leadership'
            ],
            'Health': [
                'health', 'medical', 'medicine', 'disease', 'virus', 'hospital',
                'doctor', 'patient', 'vaccine', 'treatment', 'drug', 'wellness',
                'fitness', 'pandemic', 'healthcare', 'clinical', 'therapy',
                'nutrition', 'exercise', 'mental health', 'public health',
                'research', 'study', 'clinical trial', 'healthcare system'
            ],
            'Science': [
                'science', 'research', 'study', 'scientists', 'discovery',
                'physics', 'chemistry', 'biology', 'space', 'nasa', 'experiment',
                'laboratory', 'theory', 'scientific', 'innovation', 'climate',
                'environment', 'energy', 'sustainability', 'quantum', 'genetics'
            ]
        }

        # Mock news database for reliable demonstration
        self.mock_articles_database = {
            'technology': [
                {
                    'title': 'New AI Model Achieves Breakthrough in Natural Language Understanding',
                    'description': 'Researchers at leading tech companies have developed a new transformer architecture that significantly improves machine comprehension of complex human language, potentially revolutionizing how we interact with AI systems.',
                    'source': {'name': 'Tech Innovations Daily'},
                    'content': 'A groundbreaking artificial intelligence model has demonstrated remarkable capabilities in natural language processing. The new architecture, based on advanced neural networks, shows significant improvements in understanding context, nuance, and complex linguistic structures. Early tests indicate the model can comprehend and generate human-like text with unprecedented accuracy. This development could transform industries ranging from customer service to content creation and education.',
                    'url': 'https://example.com/ai-breakthrough'
                },
                {
                    'title': 'Python 4.0 Release Announcement Expected Next Month',
                    'description': 'The Python Software Foundation has hinted at major performance improvements and new features in the upcoming Python 4.0 release, with developers eagerly awaiting the official announcement.',
                    'source': {'name': 'Developer Weekly'},
                    'content': 'After years of development, Python 4.0 is set to be unveiled next month with significant performance enhancements and new programming features. The update focuses on better parallel processing capabilities, improved memory management, and enhanced type system. Early benchmarks show up to 40% performance improvement in data processing tasks. The release also introduces new syntax features that make Python more expressive while maintaining its famous readability.',
                    'url': 'https://example.com/python-4-release'
                }
            ],
            'business': [
                {
                    'title': 'Global Stock Markets Reach Record High Amid Economic Recovery',
                    'description': 'Major stock indices worldwide have surged to all-time highs as economic indicators show strong recovery signals and corporate earnings exceed expectations across multiple sectors.',
                    'source': {'name': 'Financial Times'},
                    'content': 'Global financial markets are experiencing unprecedented growth with major indices breaking historical records. The surge is driven by better-than-expected corporate earnings, positive economic data, and optimistic growth projections. Technology and healthcare sectors are leading the gains, while traditional industries also show strong performance. Economic analysts attribute this growth to successful economic policies and rapid adaptation to changing market conditions.',
                    'url': 'https://example.com/stock-market-high'
                },
                {
                    'title': 'Startup Secures $50 Million in Series B Funding Round',
                    'description': 'An innovative tech startup focused on sustainable energy solutions has successfully closed a $50 million funding round led by prominent venture capital firms, signaling strong investor confidence in green technology.',
                    'source': {'name': 'Venture Capital News'},
                    'content': 'The startup, which develops advanced battery storage technology, has attracted significant investment from top-tier venture capital firms. The funding will accelerate product development and expand manufacturing capabilities. Company executives stated that this investment validates their business model and technological approach. The growing interest in sustainable energy solutions continues to drive substantial investment in the sector.',
                    'url': 'https://example.com/startup-funding'
                }
            ],
            'health': [
                {
                    'title': 'Breakthrough in Cancer Treatment Shows Promising Results',
                    'description': 'Clinical trials for a novel cancer therapy have demonstrated remarkable success rates, with patients showing significant improvement and minimal side effects compared to traditional treatments.',
                    'source': {'name': 'Medical Research Journal'},
                    'content': 'A revolutionary cancer treatment approach has shown extraordinary results in recent clinical trials. The therapy, which combines targeted drug delivery with immunotherapy, has achieved unprecedented success rates in treating advanced-stage cancers. Patients participating in the trials have experienced significant tumor reduction with minimal adverse effects. Medical researchers are optimistic that this approach could transform cancer treatment protocols worldwide.',
                    'url': 'https://example.com/cancer-breakthrough'
                },
                {
                    'title': 'New Study Reveals Benefits of Mediterranean Diet on Longevity',
                    'description': 'Comprehensive research spanning a decade confirms that adherence to Mediterranean dietary patterns significantly reduces risk of chronic diseases and increases life expectancy.',
                    'source': {'name': 'Health and Wellness Review'},
                    'content': 'A major longitudinal study has provided compelling evidence for the health benefits of the Mediterranean diet. The research, involving over 10,000 participants across multiple countries, shows clear correlations between this dietary pattern and reduced incidence of heart disease, diabetes, and cognitive decline. Participants following the diet most closely showed significantly longer life expectancy and better quality of life in later years. The findings reinforce the importance of dietary choices in preventive healthcare.',
                    'url': 'https://example.com/mediterranean-diet'
                }
            ],
            'science': [
                {
                    'title': 'NASA Discovers Earth-like Exoplanet in Habitable Zone',
                    'description': 'Astronomers using advanced space telescopes have identified a potentially habitable exoplanet with Earth-like characteristics, raising exciting possibilities for extraterrestrial life.',
                    'source': {'name': 'Space Exploration Daily'},
                    'content': 'In a significant astronomical discovery, NASA scientists have confirmed the existence of an Earth-like exoplanet located within the habitable zone of its star. The planet, designated Kepler-452f, shows characteristics that could support liquid water and potentially life. Detailed observations reveal an atmosphere containing oxygen and methane, strong indicators of biological activity. This discovery marks a major milestone in the search for habitable worlds beyond our solar system.',
                    'url': 'https://example.com/exoplanet-discovery'
                },
                {
                    'title': 'Quantum Computing Breakthrough Achieves New Stability Record',
                    'description': 'Researchers have demonstrated sustained quantum coherence for unprecedented durations, overcoming a major obstacle in practical quantum computing development.',
                    'source': {'name': 'Advanced Science Journal'},
                    'content': 'A team of physicists has achieved a groundbreaking milestone in quantum computing by maintaining quantum coherence for record-breaking time periods. This development addresses one of the fundamental challenges in building practical quantum computers - the fragile nature of quantum states. The new approach uses innovative error correction techniques and novel materials to protect quantum information from environmental interference. This advancement brings us closer to realizing the full potential of quantum computing for solving complex problems.',
                    'url': 'https://example.com/quantum-breakthrough'
                }
            ]
        }

    def fetch_news(self, query="artificial intelligence", page_size=5):
        """
        Fetch mock news articles based on query.

        Args:
            query (str): Search query for news articles
            page_size (int): Number of articles to fetch

        Returns:
            list: List of article dictionaries
        """
        print(f" Fetching {page_size} news articles about '{query}'...")

        # Normalize query to match categories
        query_lower = query.lower()
        matched_articles = []

        # Find articles matching the query
        for category, articles in self.mock_articles_database.items():
            if any(keyword in query_lower for keyword in category.split()):
                matched_articles.extend(articles)

        # If no specific match, mix articles from all categories
        if not matched_articles:
            for articles in self.mock_articles_database.values():
                matched_articles.extend(articles)

        # Add publication dates
        for i, article in enumerate(matched_articles):
            days_ago = random.randint(1, 30)
            article['publishedAt'] = (datetime.now() - timedelta(days=days_ago)).isoformat() + 'Z'

        # Return requested number of articles
        articles_to_return = matched_articles[:page_size]
        print(f" Successfully fetched {len(articles_to_return)} articles")
        return articles_to_return

    def preprocess_text(self, text):
        """
        Clean and preprocess text for analysis.

        Args:
            text (str): Input text to clean

        Returns:
            str: Cleaned text
        """
        if not text:
            return ""

        # Remove extra whitespace and limit length for processing
        cleaned = ' '.join(str(text).split())
        return cleaned[:2000]  # Limit text length for processing efficiency

    def summarize_article(self, text):
        """
        Generate a summary of the article text using extractive summarization.

        Args:
            text (str): Article text to summarize

        Returns:
            str: Generated summary
        """
        # Preprocess the text
        clean_text = self.preprocess_text(text)

        if not clean_text or len(clean_text.split()) < 15:
            return " Text too short for meaningful summary"

        try:
            # Simple extractive summarization by taking first few sentences
            sentences = clean_text.split('. ')
            if len(sentences) <= self.summary_sentences:
                summary = clean_text
            else:
                summary = '. '.join(sentences[:self.summary_sentences]) + '.'

            return summary if summary else " No summary could be generated"

        except Exception as e:
            return f" Summary generation failed: {str(e)}"

    def categorize_article(self, article):
        """
        Categorize article based on keyword analysis.

        Args:
            article (dict): Article data with title and description

        Returns:
            tuple: (category, confidence_score)
        """
        # Combine title and description for analysis
        title = article.get('title', '')
        description = article.get('description', '')

        if not title and not description:
            return 'General', 0

        text_to_analyze = f"{title} {description}".lower()

        # Calculate scores for each category
        scores = {}
        for category, keywords in self.categories.items():
            score = 0
            for keyword in keywords:
                if keyword in text_to_analyze:
                    score += 1
            scores[category] = score

        # Find best matching category
        best_category = max(scores, key=scores.get)
        best_score = scores[best_category]

        # Calculate confidence (0-100%)
        confidence = min(100, (best_score / len(self.categories[best_category])) * 100) if best_score > 0 else 0

        return best_category if best_score > 0 else 'General', round(confidence, 1)

    def display_article(self, article, index):
        """
        Display formatted article information.

        Args:
            article (dict): Article data
            index (int): Article number
        """
        print(f"\n{'='*80}")
        print(f" ARTICLE {index}")
        print(f"{'='*80}")

        # Title
        title = article.get('title', 'No title available')
        print(f" Title: {title}")

        # Source
        source = article.get('source', {}).get('name', 'Unknown source')
        print(f" Source: {source}")

        # URL
        url = article.get('url', 'No URL available')
        print(f" URL: {url}")

        # Category with confidence
        category, confidence = self.categorize_article(article)
        print(f"  Category: {category} (Confidence: {confidence:.1f}%)")

        # Summary
        content = article.get('content') or article.get('description') or title
        summary = self.summarize_article(content)
        print(f" Summary: {summary}")

        # Publication date
        published_at = article.get('publishedAt', 'Unknown date')
        print(f" Published: {published_at}")

    def run_demo(self, query="artificial intelligence", article_count=5):
        """
        Run a complete demonstration of the intelligent agent.

        Args:
            query (str): Search query for news
            article_count (int): Number of articles to process
        """
        print("\n" + "=" * 40)
        print("      INTELLIGENT NEWS AGENT - DEMONSTRATION")
        print("=" * 40)

        # Fetch articles
        articles = self.fetch_news(query, article_count)

        if not articles:
            print(" No articles found.")
            return

        # Process and display each article
        print(f"\n PROCESSING {len(articles)} ARTICLES...")

        category_stats = {}
        for i, article in enumerate(articles, 1):
            self.display_article(article, i)
            category, _ = self.categorize_article(article)
            category_stats[category] = category_stats.get(category, 0) + 1

        # Display summary statistics
        self.display_statistics(category_stats, len(articles))

    def display_statistics(self, category_stats, total_articles):
        """
        Display processing statistics.

        Args:
            category_stats (dict): Category frequency counts
            total_articles (int): Total number of articles processed
        """
        print(f"\n{'' * 40}")
        print("               PROCESSING STATISTICS")
        print(f"{'=' * 40}")
        print(f" Total Articles Processed: {total_articles}")
        print("\n Category Distribution:")

        for category, count in category_stats.items():
            percentage = (count / total_articles) * 100
            print(f"   • {category}: {count} articles ({percentage:.1f}%)")

        print(f"\n Demonstration completed successfully!")
        print(f" This demo used mock data to showcase full agent functionality")


class TestIntelligentNewsAgent(unittest.TestCase):
    """
    Comprehensive unit tests for the IntelligentNewsAgent class.
    """

    def setUp(self):
        """Set up test fixtures."""
        self.agent = IntelligentNewsAgent()

    def test_categorization_technology(self):
        """Test technology article categorization."""
        tech_article = {
            'title': 'New Python AI Framework Released',
            'description': 'A new machine learning library for Python developers has been launched.'
        }
        category, confidence = self.agent.categorize_article(tech_article)
        self.assertEqual(category, 'Technology')
        self.assertGreater(confidence, 0)

    def test_categorization_business(self):
        """Test business article categorization."""
        business_article = {
            'title': 'Stock Market Reaches Record High',
            'description': 'Company profits surge as economy grows.'
        }
        category, confidence = self.agent.categorize_article(business_article)
        self.assertEqual(category, 'Business')
        self.assertGreater(confidence, 0)

    def test_categorization_health(self):
        """Test health article categorization."""
        health_article = {
            'title': 'Breakthrough in Medical Research',
            'description': 'New vaccine shows promising results in clinical trials.'
        }
        category, confidence = self.agent.categorize_article(health_article)
        self.assertEqual(category, 'Health')
        self.assertGreater(confidence, 0)

    def test_categorization_general(self):
        """Test article with no specific category."""
        general_article = {
            'title': 'Beautiful Weather Today',
            'description': 'The sun is shining and birds are singing.'
        }
        category, confidence = self.agent.categorize_article(general_article)
        self.assertEqual(category, 'General')
        self.assertEqual(confidence, 0)

    def test_summarization_short_text(self):
        """Test summarization with very short text."""
        short_text = "This is a very short article."
        summary = self.agent.summarize_article(short_text)
        self.assertIn("too short", summary.lower())

    def test_summarization_valid_text(self):
        """Test summarization with adequate text."""
        long_text = """
        Artificial intelligence is transforming modern industry. Machine learning algorithms
        can now process vast amounts of data and identify patterns that were previously
        undetectable. Companies across various sectors are investing heavily in AI research
        and development. The technology shows promise in healthcare, finance, and
        transportation. However, ethical considerations around AI implementation remain
        an important topic of discussion among researchers and policymakers.
        """
        summary = self.agent.summarize_article(long_text)
        self.assertTrue(len(summary) > 20)

    def test_preprocess_text(self):
        """Test text preprocessing functionality."""
        dirty_text = "   This   is   a   test.   " * 10
        clean_text = self.agent.preprocess_text(dirty_text)
        self.assertLess(len(clean_text), len(dirty_text))
        self.assertTrue(len(clean_text) > 0)

    def test_fetch_news(self):
        """Test news fetching functionality."""
        articles = self.agent.fetch_news("technology", 2)
        self.assertEqual(len(articles), 2)
        self.assertIn('title', articles[0])


def run_tests():
    """Run all unit tests and display results."""
    print("\n" + "=" * 40)
    print("              RUNNING UNIT TESTS")
    print("=" * 40)

    # Create test suite
    loader = unittest.TestLoader()
    suite = loader.loadTestsFromTestCase(TestIntelligentNewsAgent)

    # Run tests
    runner = unittest.TextTestRunner(verbosity=2)
    result = runner.run(suite)

    # Display test summary
    print(f"\n TEST SUMMARY:")
    print(f"   • Tests Run: {result.testsRun}")
    print(f"   • Failures: {len(result.failures)}")
    print(f"   • Errors: {len(result.errors)}")
    print(f"   • Success: {result.wasSuccessful()}")

    return result.wasSuccessful()


def main():
    """
    Main function with interactive menu system.
    """
    # Initialize agent
    agent = IntelligentNewsAgent()

    while True:
        print("\n" + "=" * 40)
        print("=        INTELLIGENT NEWS AGENT - MAIN MENU")
        print("=" * 40)
        print("1.  Run Full Demonstration")
        print("2.  Search Specific Topic")
        print("3.  Run Unit Tests")
        print("4.  Agent Information")
        print("5.  Exit")

        choice = input("\nEnter your choice (1-5): ").strip()

        if choice == '1':
            # Run default demonstration
            agent.run_demo("artificial intelligence", 5)

        elif choice == '2':
            # Custom search
            topic = input("Enter search topic: ").strip() or "technology"
            try:
                count = int(input("Number of articles (1-10): ") or "5")
                count = max(1, min(10, count))  # Limit between 1-10
            except ValueError:
                count = 5

            agent.run_demo(topic, count)

        elif choice == '3':
            # Run tests
            run_tests()

        elif choice == '4':
            # Display agent information
            print(f"\n AGENT INFORMATION:")
            print(f"   • Categories: {', '.join(agent.categories.keys())}")
            print(f"   • Summary Sentences: {agent.summary_sentences}")
            print(f"   • Language: {agent.language}")
            print(f"   • Data Source: Mock Database (No API key required)")
            print(f"   • Total Mock Articles: {sum(len(articles) for articles in agent.mock_articles_database.values())}")

        elif choice == '5':
            print(" Thank you for using the Intelligent News Agent!")
            break

        else:
            print(" Invalid choice. Please enter 1-5.")

        input("\nPress Enter to continue...")


if __name__ == "__main__":
    # Check if running in test mode
    if len(sys.argv) > 1 and sys.argv[1] == 'test':
        run_tests()
    else:
        main()
```
